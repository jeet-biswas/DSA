2.1.1 Pair Sum in a Sorted Array

You are given a sorted array and a target.Find two elements whose sum is 10.

```python
arr = [1, 1, 1, 4, 4, 4]
target = 5
left=0
current_sum=0
right=len(arr)-1
while left<right:
    current_sum=arr[left]+arr[right]
    if current_sum==target:
        print(arr[left],arr[right])
        left+=1
        right-=1
    elif current_sum>target:
        right-=1
    else:
        left+=1
```

2.1.2: Palindrome / Symmetric Comparison


Write Python code to determine whether the array is a palindrome using the opposite-direction Two Pointer approach.

```python
arr = [1, 2, 3, 4, 3, 2, 1]
left=0
right=len(arr)-1
flag=True
while left<=right:
    if arr[left]!=arr[right]:
        flag=False
        break
    left+=1
    right-=1
flag
```

2.1.3: Reverse / Swap From Both Ends

"Reverse the array in-place."

```python
arr = [10, 20, 30, 40, 50, 60]
left=0
right=len(arr)-1
while left<right:
    arr[left],arr[right]=arr[right],arr[left]
    right-=1
    left+=1
arr
```

2.1.4: Partition / Boundary

Move all elements smaller than 5 to the left and elements greater than or equal to 5 to the right.

```python
arr = [1, 5, 2, 8, 3, 4]
left = 0
right = len(arr) - 1
while left < right:
    while left < right and arr[left] < 5:
        left += 1
    while left < right and arr[right] >= 5:
        right -= 1
    if left < right:
        arr[left], arr[right] = arr[right], arr[left]
print(arr)
```

2.1.5: Maximum / Minimum Between Two Ends

Container With Most Water.

```python
arr = [1, 8, 6, 2, 5, 4, 8, 3, 7]
left=0
right=len(arr)-1
maxi=float("-inf")
while left<right:
    width=right-left
    area=width*min(arr[left],arr[right])
    maxi=max(area,maxi)
    if arr[left]<arr[right]:
        left+=1
    else:
        right-=1
maxi
```

2.1.6: Sorted Array + Duplicates / Constraints
