3.1 Running sum 


Create a new array containing the running sum.

```python
arr = [5, 2, 8, 3, 6]
prefix_array=[0]*len(arr)
prefix_array[0]=arr[0]
for i in range(1,len(arr)):
    prefix_array[i]=arr[i]+prefix_array[i-1]
prefix_array

# another technique
arr = [5, 2, 8, 3, 6]
new_arr=[]
result=0
for i in arr:
    result+=i
    new_arr.append(result)
new_arr
```
