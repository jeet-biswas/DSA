3.1 Running sum 


Create a new array containing the running sum.

```python
arr = [5, 2, 8, 3, 6]
prefix_array=[0]*len(arr)
prefix_array[0]=arr[0]
for i in range(1,len(arr)):
    prefix_array[i]=arr[i]+prefix_array[i-1]
prefix_array

# another technique
arr = [5, 2, 8, 3, 6]
new_arr=[]
result=0
for i in arr:
    result+=i
    new_arr.append(result)
new_arr
```

3.2 — Prefix Sum for Range Sum Queries

Find the sum of elements from index 2 to index 5 without using a loop.

```python
arr = [6, 3, 8, 2, 5, 4,6]
queries = [
    [1, 3],
    [0, 4],
    [2, 6],
    [4, 5]
]
prefix_array=[0]*len(arr)
prefix_array[0]=arr[0]
for i in range(1,len(arr)):
    prefix_array[i]=arr[i]+prefix_array[i-1]
for L,R in queries:
    if L==0:
        print(prefix_array[R])
    else:
        print(prefix_array[R]-prefix_array[L-1])    
```

3.5 — Code: Subarray Sum = K

```python
arr = [2, 5, 3, 7, 1]
K = 10

prefix = [0] * (len(arr) + 1)
for i in range(len(arr)):
    prefix[i+1] = prefix[i]+arr[i]
found = False
for L in range(len(arr)):
    for R in range(L, len(arr)):
        current_sum = prefix[R + 1] - prefix[L]
        if current_sum == K:
            print("Subarray found")
            print("L =", L)
            print("R =", R)
            print("Subarray =", arr[L:R + 1])
            found = True
if not found:
    print("No subarray found")
```

3.6 — Prefix Sum for Counting / Frequency of Subarray Sums

Count how many contiguous subarrays have a sum exactly equal to K.

```python
arr = [1, 2, 1, 2, 1]
K = 3
count=0
prefix_sum=0 
freq={0:1} #because before we process any element, the prefix sum is 0, and it has occurred once.
for i in range(len(arr)):
    prefix_sum+=arr[i]
    needed=prefix_sum-K
    if needed in freq:
        count+=freq[needed]
    if prefix_sum in freq:
        freq[prefix_sum]+=1
    else:
        freq[prefix_sum]=1
count
```

3.8 — Longest Zero-Sum Subarray

```python
arr = [3, 4, -7, 2, -2]
longest=0
prefix_sum=0
freq={0:-1} # we use -1 because prefix sum is 0 before array started
for i in range(len(arr)):
    prefix_sum+=arr[i]
    if prefix_sum in freq:
        length=i-freq[prefix_sum]
        longest=max(longest,length)
    else:
        freq[prefix_sum]=i
longest
```
